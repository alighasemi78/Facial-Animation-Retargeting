## Installing PyTorch3D

```python
import os
import sys
import torch
import subprocess
need_pytorch3d=False
try:
    import pytorch3d
except ModuleNotFoundError:
    need_pytorch3d=True
if need_pytorch3d:
    pyt_version_str=torch.__version__.split("+")[0].replace(".", "")
    version_str="".join([
        f"py3{sys.version_info.minor}_cu",
        torch.version.cuda.replace(".",""),
        f"_pyt{pyt_version_str}"
    ])
    get_ipython().system('pip install -q iopath')
    if sys.platform.startswith("linux"):
        print("Trying to install wheel for PyTorch3D")
        get_ipython().system('pip install -q --no-index --no-cache-dir pytorch3d -f https://dl.fbaipublicfiles.com/pytorch3d/packaging/wheels/{version_str}/download.html')
        pip_list = get_ipython().getoutput('pip freeze')
        need_pytorch3d = not any(i.startswith("pytorch3d==") for  i in pip_list)
    if need_pytorch3d:
        print(f"failed to find/install wheel for {version_str}")
if need_pytorch3d:
    print("Installing PyTorch3D from source")
    get_ipython().system('pip install -q ninja')
    get_ipython().system("pip install -q 'git+https://github.com/facebookresearch/pytorch3d.git@stable'")
```
